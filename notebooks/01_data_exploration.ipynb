# Data Exploration for Image Classification

This notebook explores datasets for image classification tasks.

## 1. Setup and Imports

```python
import sys
import os
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.datasets import cifar10, fashion_mnist

import config
from utils.data_loader import DataLoader
from utils.visualization import plot_class_distribution

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

print("Setup complete!")
```

## 2. Load Dataset

```python
# Choose dataset: 'cifar10', 'fashion_mnist', or 'mnist'
dataset_name = 'cifar10'

# Load data
loader = DataLoader(dataset_name=dataset_name)
(x_train, y_train), (x_val, y_val), (x_test, y_test), class_names = loader.load_builtin_dataset()

print(f"\nDataset: {dataset_name}")
print(f"Training samples: {len(x_train)}")
print(f"Validation samples: {len(x_val)}")
print(f"Test samples: {len(x_test)}")
print(f"Image shape: {x_train.shape[1:]}")
print(f"Number of classes: {len(class_names)}")
print(f"Class names: {class_names}")
```

## 3. Visualize Sample Images

```python
# Display random samples
num_samples = 16
indices = np.random.choice(len(x_train), num_samples, replace=False)

fig, axes = plt.subplots(4, 4, figsize=(12, 12))
axes = axes.flatten()

for idx, ax in zip(indices, axes):
    img = x_train[idx]
    label = np.argmax(y_train[idx])
    
    if img.shape[-1] == 1:
        ax.imshow(img.squeeze(), cmap='gray')
    else:
        ax.imshow(img)
    
    ax.set_title(f'{class_names[label]}', fontsize=10)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 4. Class Distribution

```python
# Plot class distribution
plot_class_distribution(y_train, class_names, title='Training Set Class Distribution')
plot_class_distribution(y_test, class_names, title='Test Set Class Distribution')
```

## 5. Image Statistics

```python
# Calculate statistics
print("Image Statistics:")
print(f"Min pixel value: {x_train.min():.2f}")
print(f"Max pixel value: {x_train.max():.2f}")
print(f"Mean pixel value: {x_train.mean():.2f}")
print(f"Std pixel value: {x_train.std():.2f}")

# Plot pixel value distribution
plt.figure(figsize=(10, 5))
plt.hist(x_train.flatten(), bins=50, alpha=0.7, color='steelblue')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.title('Pixel Value Distribution')
plt.grid(alpha=0.3)
plt.show()
```

## 6. Visualize One Image Per Class

```python
# Find one example of each class
num_classes = len(class_names)
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()

for class_idx in range(num_classes):
    # Find first occurrence of this class
    idx = np.where(np.argmax(y_train, axis=1) == class_idx)[0][0]
    img = x_train[idx]
    
    if img.shape[-1] == 1:
        axes[class_idx].imshow(img.squeeze(), cmap='gray')
    else:
        axes[class_idx].imshow(img)
    
    axes[class_idx].set_title(f'{class_names[class_idx]}', fontsize=12, fontweight='bold')
    axes[class_idx].axis('off')

plt.tight_layout()
plt.show()
```

## 7. Data Augmentation Preview

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Create data generator with augmentation
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    zoom_range=0.2
)

# Select one image
sample_img = x_train[0:1]

# Generate augmented versions
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
axes = axes.flatten()

# Original image
if sample_img.shape[-1] == 1:
    axes[0].imshow(sample_img[0].squeeze(), cmap='gray')
else:
    axes[0].imshow(sample_img[0])
axes[0].set_title('Original', fontweight='bold')
axes[0].axis('off')

# Augmented versions
i = 1
for batch in datagen.flow(sample_img, batch_size=1):
    if batch.shape[-1] == 1:
        axes[i].imshow(batch[0].squeeze(), cmap='gray')
    else:
        axes[i].imshow(batch[0])
    axes[i].set_title(f'Augmented {i}', fontweight='bold')
    axes[i].axis('off')
    i += 1
    if i >= 8:
        break

plt.tight_layout()
plt.show()
```

## 8. Summary

```python
print("\n" + "="*80)
print("DATASET SUMMARY")
print("="*80)
print(f"Dataset: {dataset_name}")
print(f"Total samples: {len(x_train) + len(x_val) + len(x_test)}")
print(f"Training: {len(x_train)} ({len(x_train)/(len(x_train)+len(x_val)+len(x_test))*100:.1f}%)")
print(f"Validation: {len(x_val)} ({len(x_val)/(len(x_train)+len(x_val)+len(x_test))*100:.1f}%)")
print(f"Test: {len(x_test)} ({len(x_test)/(len(x_train)+len(x_val)+len(x_test))*100:.1f}%)")
print(f"\nImage dimensions: {x_train.shape[1:]}")
print(f"Number of classes: {len(class_names)}")
print(f"Classes: {', '.join(class_names)}")
print("="*80)
```
